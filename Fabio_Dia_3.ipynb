# As Condicionais ` if ` e ` else `

Uma situação comum nos primeiros meses do ano na capital paraense é a chuva matinal. E é bem provável que já tenha se deparado com ela desde o início do Nivelamento. A partir disso, temos uma pequena decisão a ser tomada todo começo de dia: trazer ou não o guarda-chuva? E isso pode depender diretamente do estado do tempo naquele momento, pela a seguinte lógica:


**Se tempo == chovendo, levo meu guarda-chuva;**

**Senão (se tempo != chovendo), não levo meu guarda-chuva.**

Já trabalhamos esse tipo de lógica nas primeiras aulas, mas, agora é importante formalizar o pensamento para a linguagem computacional, que é feita da seguinte forma:

Se --> if

Senão --> else

Na linguagem de programação python teríamos:

```python
tempo = "chuvoso"

if tempo == "chuvoso":
  print("Levo meu guarda-chuva")
else:
  print("Não levo meu guarda-chuva.")
```

Além disso, é comum que uma condição precise ser mais detalhada. Por exemplo, pode ser necessário diferenciar entre vários tipos de clima — como um dia nublado ou ensolarado. Nesse caso, usamos o ` elif ` (abreviação de "else if") para criar múltiplos caminhos na decisão:

```python
if tempo == "chovendo":
    print("Levo meu guarda-chuva.")
elif tempo == "nublado":
    print("Melhor levar um casaco leve.")
else:
    print("Dia ensolarado, não preciso de guarda-chuva.")
```

Essa estrutura permite que o programa escolha entre vários cenários, garantindo que cada situação seja tratada de forma adequada.

Em resumo
- if:  Usamos para a primeira condição.
- elif: quando há mais de uma condição a ser verificada.
- else: para cobrir todos os outros casos não mencionados.

##Operadores de comparação

Para tomar decisões na programação já sabemos que devemos usar ` if ` e e ` else `, mas para usa-los precisamos comparar valores e estados. É aqui que os operadores de comparação entram em cena. Eles nos permitem testar condições e ver se determinadas afirmações são verdadeiras ou falsas. Relembrando os operadores de comparação:


<hr>

| Operador | Descrição                         | Exemplo   |
|----------|-----------------------------------|-----------|
| ==       | Igualdade                         | 4 == 5 #False |
| !=       | Diferença                         | 4 != 5 #True  |
| >        | Maior que                         | 4 > 5 #False  |
| <        | Menor que                         | 4 < 5 #True   |
| >=       | Maior ou igual que                | 4 >= 5 #False |
| <=       | Menor ou igual que                | 4 <= 5 #True  |

Igualdade (==): Verifica se dois valores são iguais.

```python
if tempo == "chuvoso":
    print("Usar guarda-chuva.")
```

Diferença (!=): Verifica se dois valores são diferentes.

```python
if tempo != "chuvoso":
    print("Não preciso de guarda-chuva.")
```

Maior que (>), Menor que (<), Maior ou igual (>=), Menor ou igual (<=):
Esses operadores comparam valores numéricos.

```python
temperatura = 40
if temperatura > 35:
  print("Está muito quente!")
else:
  print("A temperatura está agradável.")
```

#Laços de repetição

Quando estamos programando, é muito comum encontrarmos situações em que um mesmo comando precisa ser executado várias vezes, até que uma condição seja atingida.

Para resolver isso, usamos os laços de repetição.

Essas estruturas permitem que uma mesma tarefa seja repetida automaticamente, sem que precisemos escrever o mesmo comando várias vezes, chamamos isso de looping. Os dois laços de repetição mais usados em Python são o while e o for.

##While

O ` while ` é um laço de repetição que executa um bloco de código enquanto uma condição for verdadeira.

```python
cont = 1
while cont <= 5:  #Repete enquanto cont for menor ou igual a 5
  print("Contagem: ", cont)
  cont = cont + 1
  #Fim do looping
print("Fim da contagem.")
```

No exemplo acima o while é usado para contar de 1 a 5. Porém, pode-se usar para outras aplicações.

```python
#Neste código é exigido uma senha de um usuário até ele acertar ela

senha="nivelamento"
tentativa=input("Digite a senha: ")
while senha!=tentativa:
  print("Senha incorreta")
  tentativa=input("Digite a senha: ")
print("Senha correta!")
```

##For

O laço de repetição ` for ` é uma forma mais "automática" de criar repetições do que o ` while `.
Com ele, não precisamos nos preocupar em criar e atualizar a variável de controle manualmente — o próprio ` for ` faz isso para nós.

```python
for i in range(1, 6):  #Repete de 1 a 5
  #Começo do looping
  print("Contagem: ", i)
  #Fim do looping
print("Fim da contagem.")
```

O código acima faz uma contagem de 1 a 5, assim como no exemplo com while.

A diferença é que com o for, não precisamos criar nem atualizar a variável de contagem (i) manualmente — o próprio laço faz isso automaticamente.

Além disso, no ` range(1, 6) `:
  - Primeiro número (1) é o valor inicial
  -  O segundo número (6) é o valor final, mas ele não é incluído na contagem
(ou seja, a contagem vai até um número antes: 5)

O for não precisar ter necessariamente um valor inicial e um final, ele pode indicar o número de vezes em que uma determinada ação será executada.

```python
#Implementação de exponenciação a partir de multiplicações

#Entrada de dados
base = int(input("Base: "))
expoente = int(input("Expoente: "))

resultado = 1 # Valor inicial

#base**expoente
for _ in range(expoente):
  resultado = resultado * base

print(resultado)
```

Perceba que o "expoente" determina apenas o número de vezes que a multiplicação ocorrerá, ou seja, o número de repetições.

Note também que a variável "i", comumente utilizada para a contagem, foi subsituida por ` _  `, porque o contador não é usado diretamente nas operações, então nessa situação pode ser desconsiderado  

O `for` também pode ser utilizado para pecorrer strings ou listas.

```python
palavra="Antonio"
for letra in palavra:
  print(letra)
```

Você também consegue utilizar o looping para encontrar um elemnto específico da lista que você está pecorrendo.

```python
palavra="Antonio"
for letra in palavra:
  if letra=="n":
    print(letra)
```

A variável ` letra ` percorre cada letra de ` palavra `, isso mostra que o laço for não serve apenas para contagem, mas pode ser utilizado para listar algo, por exemplo.

##Quando usar cada um

Use ` while ` quando:
- A repetição depende de uma condição que pode mudar de forma menos previsível, ou seja, não é possível determinar previamente o número exato de iterações.
- Você precisa continuar o laço até que uma determinada condição seja satisfeita (por exemplo, ler a entrada do usuário até que ele digite a senha correta).

Use ` for ` quando:
- Você sabe exatamente quantas vezes a tarefa deve ser repetida.
- O loop é mais “automático”, já que não precisa cuidar da atualização da variável de controle.

##Break

O break é um comando especial usado em laços de repetição (` for ` e ` while ` ) para interromper imediatamente a execução do loop, mesmo que a condição de parada ainda não tenha sido atingida.

Isso é útil quando o loop atingi seu objeivo antes da condição de parada.

```python
#Implementação da senha com número limitado de tentativas

senha = "looping"
tentativas = 5
correto = False

for _ in range(tentativas):
  palavra = input("Digite a senha: ")
  if palavra!=senha:
    print("Senha incorreta!")
  else:
    correto = True
    break

if correto:
  print("senha correta!")
else:
  print("Número máximo de tentativas alcançado")
```

No exemplo acima o usário tem um número determinados de chances de acertar a senha, o looping está programado repetir até acabarem as chances. Porém se ele acertar a senha antes das chances acabarem o looping é quebrado com o ` break `

Instruções Passo a Passo:
1. Primeiro Número:
  - O programa deve pedir ao usuário para digitar o primeiro número.
2. Mostrar um menu com as opções numeradas
  - 1 → Adição (+)
  - 2 → Subtração (-)
  - 3 → Multiplicação (*)
  - 4 → Divisão (/)
  - 5 → Finalizar e mostrar resultado
3. Escolha da Operação:
 - O usuário deve digitar o número da operação que deseja realizar (1, 2, 3, 4 ou 5).
4. Segundo Número (se não for finalizar):
 - Se o usuário não escolher 5 (Finalizar), o programa deve pedir um segundo número.
5. Cálculo e Atualização do Resultado:
 - O programa realiza a operação escolhida entre o resultado atual e o novo número.
 - O resultado vira o primeiro número da próxima operação.
6. Repetição:
 - O programa volta para o menu de operações, permitindo que o usuário continue calculando ou finalize.
7.Finalização:
 - Quando o usuário escolher 5 (Finalizar), o programa mostra o resultado final e encerra.
