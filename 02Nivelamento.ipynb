```python
#Variáveis
v1 = int(input("Digite um valor: "))
v2 = int(input("Digiter outro valor: "))
#Operações
print(v1==v2)
print(v1!=v2)
print(v1>v2)
print(v1<v2)
print(v1>=v2)
print(v1<=v2)
#Operações com lógica proposicional
print(f"{v1 >= 6 and v2 <=7} ")
print(f"{v1 > 8 or v2 != 10}")
print(f"{not(v1<v2)}")
```
