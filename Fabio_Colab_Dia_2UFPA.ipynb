##Variáveis

Assim como na matemática, variáveis, em programação, representam valores que podem ser armazenados e manipulados ao longo do programa.

Esses valores podem ser números, textos, letras, frases e muito mais. O Python reconhece o tipo de dado armazenado e o classifica automaticamente.

###int: Números inteiros

Usado para representar números sem parte decimal.

```python
idade = 25
print("Idade:", idade)
print(type(idade))  #Aqui é identificado o tipo da variável
```

###float: Número decimal (ponto flutuante)

Usado quando precisamos de números com parte decimal.

```python
altura = 1.75
print("Altura:", altura)
print(type(altura))
```

###str: Texto(string)

Tudo que estiver entre aspas (simples ou duplas) é considerado texto, em programação é usualmente nomeado de string

```python
nome = "Desirée"
print("Nome:", nome)
print(type(nome))
```

###Bool: Valor lógico(booleano)

Usado para armazenar verdadeiro ou falso (útil em decisões e comparações).

```python
programando = False
print("Está programando?", programando)
print(type(programando))
```

## Operações

Programas fazem uso das operações matemáticas de diversas formas, a línguagem Python traz suporte para múltiplas operações como mostradas na tabela a seguir, a coluna "Operador" mostra o que deve ser usado para utilizar essas operações.

<hr>

| Operador | Descrição                         | Exemplo   |
|----------|-----------------------------------|-----------|
| +        | Adição                            | 5 + 3 = 8 |
| -        | Subtração                         | 5 - 3 = 2 |
| *        | Multiplicação                     | 5 * 3 = 15|
| /        | Divisão                           | 5 / 3 ≈ 1.6667|
| //       | Divisão inteira                   | 5 // 3 = 1|
| %        | Módulo (resto da divisão)         | 5 % 3 = 2 |
| **       | Exponenciação                     | 5 ** 3 = 125 |


```python
print("Escolha dois números")
primeiro_valor = int(input("Digite um número "))
segundo_valor = int(input("Digite outro número "))

print(primeiro_valor + segundo_valor)
print(primeiro_valor - segundo_valor)
print(primeiro_valor * segundo_valor)
print(primeiro_valor % segundo_valor)
print(primeiro_valor ** segundo_valor)
```

### Ordem de operações

Um conceito muito importante para programação como para a matemática cotidiana é a ordem em que operações são efetuadas se há mais de uma operação a ser executada, para python pode-se ter algo como:

    valor = input()
    valor = valor + 2 * valor - 4 / 5

Em circunstâncias assim Python dispõe de uma ordem predefinida em que executa as operações:

1. Parênteses ( )

2. Exponenciação.

3. Multiplicação, Divisão, Divisão Inteira e Módulo

4. Adição e Subtração

Python segue essa ordem e da esquerda para a direita.

No exemplo a seguir há duas expressões, original e alterado, executar a célula mostra como a presença dos parenteses muda o resultado dando prioridade a um parte da expressão.


```python
original = (1 + 2) * (3 - 4) / 5

alterado = 1 + 2 * 3 - 4 / 5

print(f"Valor original: {original} ")

print(f"Valor alterado: {alterado}")
```

# Operadores Relacionais

Os operadores relacionais são utilizados em operações que retornam um valor booleano, dado uma expressão de comparação (<, >, ==), se a expressão for verdadeira, é retornado `True`, se não, é retornado `False`.

Essas operações serão muito importantes na próxima aula, quando veremos estruturas condicionais e de repetição.

```python
n = int(input("digite um numero: "))
print("o número está entre 5 e 10? ", n >= 5 and n <= 10)
```

```python
print("Escolha dois números")
primeiro_valor = int(input("Digite um número "))
segundo_valor = int(input("Digite outro número "))

print(primeiro_valor == segundo_valor)
print(primeiro_valor != segundo_valor)
print(primeiro_valor > segundo_valor)
print(primeiro_valor < segundo_valor)
print(primeiro_valor <= segundo_valor)
print(primeiro_valor >= segundo_valor)
```
